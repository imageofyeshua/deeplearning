```python
# import libraries
import numpy as np
import torch
import torch.nn as nn
```

```python
# create a vector
v = np.array([ 1,40,2,-3 ])

# find and report the maximum and minimum values
minval = np.min(v)
maxval = np.max(v)

print('Min: %g, Max: %g' %(minval,maxval) )

# now for the argmin/argmax
minidx = np.argmin(v)
maxidx = np.argmax(v)

print('Min index: %g, Max index: %g' %(minidx,maxidx) )

# confirm
print(f'Min val is { v[minidx] }, max val is { v[maxidx] }')
```

```python
# repeat with matrix
M = np.array([ [0,1,10], 
               [20,8,5] ])
print(M)
print()

# various minima in this matrix!
minvals1 = np.min(M)        # minimum from Entire matrix
minvals2 = np.min(M,axis=0) # minimum of each column (across rows)
minvals3 = np.min(M,axis=1) # minimum of each row (across columns)

# print them out
print(minvals1)
print(minvals2)
print(minvals3)
```
